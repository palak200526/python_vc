```python
import pandas as pd
import joblib
import os

from feature_engineering import ChurnFeatureEngineer
```

# Task 1 — Load the Saved Model

The first task is to load the trained machine learning Pipeline created in Notebook 1.

The model is stored as a `.pkl` file using Joblib. Before loading it, the notebook checks whether the model artifact exists at the expected location.

The complete Pipeline is loaded rather than rebuilding the preprocessing or training steps.

### Workflow

**Check Model File → Load Saved Pipeline → Verify Model Type**

```python
model_path = "model/final_churn_model.pkl"

print("Model exists:", os.path.exists(model_path))
```

###  Load Final Model

This code loads the **saved churn prediction model** using `joblib` and displays its model type to confirm that it was loaded successfully.


```python
final_model = joblib.load(model_path)

print("Final churn model loaded successfully!")
print(type(final_model))
```

# Task 2 — Prepare New / Raw Data

The second task is to prepare customer records for prediction.

The data is loaded from the available dataset in the `data` folder. The raw customer records contain both input features and columns that should not be provided to the model, such as customer identifiers and churn-related target information.

Only the original input features expected by the saved Pipeline are used for prediction.

Importantly, the preprocessing and feature engineering are **not manually recreated** in this notebook. The saved Pipeline performs those transformations automatically during prediction.

### Workflow

**Identify Data File → Load Raw Data → Remove Unnecessary/Target Columns → Verify Model Input Features → Prepare Prediction Data**

###  List Data Files

This code lists and displays all the **files present in the `data` folder**, helping identify the available datasets for prediction.


```python
data_files = os.listdir("data")
print(data_files)
```

###  Load Dataset for Prediction

This code loads the **Telco Customer Churn Excel dataset** into a DataFrame and displays its **shape and first few rows** to verify that the data was loaded correctly.


```python
data_path = "data/Telco_customer_churn.xlsx"

df_new = pd.read_excel(data_path)

print("Data loaded successfully!")
print("Shape:", df_new.shape)
print(df_new.head())
```

###  Prepare Prediction Data

This code removes **unnecessary and target-related columns** from the new dataset and creates `X_new` containing only the **features required for prediction**.


```python
X_new = df_new.drop(
    columns=[
        "Churn Label",
        "Churn Value",
        "Churn Score",
        "Churn Reason"
    ],
    errors="ignore"
)

print("Raw prediction data shape:", X_new.shape)
print("Raw input columns:")
print(X_new.columns.tolist())
```

###  Check Model Features

This code retrieves the **input features expected by the final model** and displays their **total count and names** to verify compatibility with the prediction data.


```python
engineered_features = final_model.named_steps["preprocessor"].feature_names_in_

print("Engineered features:", len(engineered_features))
print(list(engineered_features))
```

###  Recover Original Input Features

This code removes the **4 engineered features** from the model's feature list to recover the **original input features** required for prediction.


```python
engineered_only = [
    "Tenure Years",
    "Average Monthly Charge",
    "Service Count",
    "Charge Tenure Interaction"
]

expected_features = [
    col for col in engineered_features
    if col not in engineered_only
]

print("Number of original input features:", len(expected_features))
print("\nOriginal input features:")
print(expected_features)
```

### Prepare Prediction Features

This code selects the **expected input features** from the new dataset, checks the **data shape and total missing values**, and displays the first few rows for verification.


```python
# X_new = df_new[expected_features].copy()
X_new = df_new[expected_features].head(10).copy()


print("Prediction data shape:", X_new.shape)
print("\nMissing values:")
print(X_new.isnull().sum().sum())

X_new.head()
```

# Task 3 — Generate Predictions

After preparing the raw input features, the saved Pipeline is used to generate predictions.

A small sample of 10 customer records is selected to demonstrate the inference process. The model predicts whether each customer is likely to churn and also provides the corresponding churn probability.

The important point is that the saved Pipeline automatically applies the same feature engineering and preprocessing steps used during training before generating the final prediction.

### Workflow

**Select Unseen Customer Records → Pass Raw Features to Saved Pipeline → Apply Stored Feature Engineering & Preprocessing → Predict Churn Class → Calculate Churn Probability → Display Results**

###  Select Prediction Sample

This code selects the **first 10 customers** from the prepared dataset for prediction and displays their feature data.


```python
X_predict = X_new.head(10).copy()

print("Prediction data shape:", X_predict.shape)
X_predict
```

###  Generate Predictions

This code uses the **final trained model** to predict whether each customer will churn and calculates the **probability of churn** for each prediction.


```python
predictions = final_model.predict(X_predict)

prediction_probabilities = final_model.predict_proba(X_predict)[:, 1]

print("Predictions completed!")
print("Number of predictions:", len(predictions))
```

## Prediction Output

The prediction output contains the original customer input records along with two additional fields:

- **Predicted Churn** — the class predicted by the trained model.
- **Churn Probability** — the model's estimated probability that the customer will churn.

This allows the prediction results to be interpreted together with the original customer information.

###  Prediction Results

This code adds the **predicted churn status** and **churn probability** to the prediction data and displays the results for the selected customers.


```python
results = X_predict.copy()

results["Predicted Churn"] = predictions
results["Churn Probability"] = prediction_probabilities.round(4)

results[[
    "Predicted Churn",
    "Churn Probability"
]]
```

###  Display Results

`display(results)` shows the **complete prediction results table** in a clear, readable format.


```python
display(results)
```

## Prediction Verification

The final verification confirms that the saved Pipeline was successfully loaded and that predictions were generated for the selected customer records.

This demonstrates the complete inference workflow without retraining the model.

###  Prediction Summary

This code prints a **confirmation message** showing that the saved model was loaded successfully, predictions were completed on **unseen customer records**, and the total number of predicted records.


```python
print("Model loaded from saved artifact:", model_path)
print("Prediction completed successfully on unseen customer records.")
print("Number of records predicted:", len(results))
```

# Task 4 — Explain the Design

## Why Save the Complete Pipeline?

Saving the complete machine learning Pipeline is safer and more reproducible than saving only the trained model.

The Pipeline keeps the feature engineering, preprocessing, and trained model together. Therefore, when new customer data is provided, the same transformations used during training can automatically be applied before prediction.

This reduces the risk of applying different preprocessing steps during inference and helps ensure that the model receives data in the same format as during training.

It also makes the model easier to reuse because the prediction notebook does not need to manually recreate the feature engineering or preprocessing logic.

### Key Benefits

- **Consistency:** The same transformations are used during training and prediction.
- **Reproducibility:** The complete workflow can be reused later.
- **Reduced errors:** Manual preprocessing differences are avoided.
- **Easy deployment:** The saved Pipeline can directly accept raw input features and produce predictions.
