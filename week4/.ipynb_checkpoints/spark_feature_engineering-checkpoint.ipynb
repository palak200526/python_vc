# Task 2 — Select Feature Engineering Transformations

The feature-engineering transformations selected for this notebook are transformations that were already implemented in Notebook 1.

Two transformations are selected:

1. **Tenure Years**
2. **Service Count**

These transformations were chosen because they create meaningful customer-level features from the existing raw customer information.

The assignment requires only 1–2 feature-engineering transformations to be translated into PySpark.

## Selected Transformations

### 1. Tenure Years

The `Tenure Months` feature is converted into `Tenure Years`.

This provides a more interpretable representation of how long the customer has been with the company.

### 2. Service Count

The service-related columns are combined to create a numerical feature representing the number of services used by each customer.

This summarizes multiple service indicators into one customer-level feature.

## Selected Feature Engineering Transformations

Two feature-engineering transformations from Notebook 1 are reimplemented using PySpark DataFrame APIs:

1. **Tenure Years** — converts customer tenure from months into years.
2. **Average Monthly Charge** — calculates the average monthly charge using Total Charges divided by Tenure Months.

The Average Monthly Charge transformation is a ratio-based derived feature, satisfying the requirement to create a derived feature.

```python
df_spark = df_spark.withColumn(
    "Tenure Years",
    col("Tenure Months") / lit(12)
)

df_spark.select(
    "Tenure Months",
    "Tenure Years"
).show(10)
```

```python
from pyspark.sql.functions import col, lit, when

df_spark = df_spark.withColumn(
    "Total Charges",
    col("Total Charges").cast("double")
)
```

```python
df_spark = df_spark.withColumn(
    "Average Monthly Charge",
    when(
        col("Tenure Months") > 0,
        col("Total Charges") / col("Tenure Months")
    ).otherwise(0)
)

df_spark.select(
    "Total Charges",
    "Tenure Months",
    "Average Monthly Charge"
).show(10)
```

# Task 4 — Derived Feature

At least one of the feature-engineering transformations creates a derived feature from existing customer attributes.

In this notebook, **Tenure Years** is a derived feature created from **Tenure Months**.

The transformation is:

$$
\text{Tenure Years} = \frac{\text{Tenure Months}}{12}
$$

This demonstrates how a new meaningful feature can be derived from existing customer data using PySpark DataFrame operations.

Another derived feature implemented is **Average Monthly Charge**, calculated using Total Charges and Tenure Months.

## Task 5 — Categorical Encoding

The `Contract` feature is a categorical feature with values such as `Month-to-month`, `One year`, and `Two year`.

In Notebook 1, categorical features were handled using One-Hot Encoding. For the Spark implementation, `StringIndexer` is used to convert the categorical values into numerical indices, followed by `OneHotEncoder` to create a numerical vector representation.

This allows the categorical `Contract` feature to be used by machine learning algorithms while preserving its categorical nature.

The encoding is implemented using PySpark ML APIs.

```python
from pyspark.ml.feature import StringIndexer, OneHotEncoder

contract_indexer = StringIndexer(
    inputCol="Contract",
    outputCol="Contract_Index",
    handleInvalid="keep"
)

df_indexed = contract_indexer.fit(df_spark).transform(df_spark)

contract_encoder = OneHotEncoder(
    inputCol="Contract_Index",
    outputCol="Contract_Encoded"
)

encoder_model = contract_encoder.fit(df_indexed)
df_encoded = encoder_model.transform(df_indexed)

df_encoded.select(
    "Contract",
    "Contract_Index",
    "Contract_Encoded"
).show(10, truncate=False)
```

## Task 8 — Comparison of Pandas/scikit-learn and PySpark Implementations

The feature-engineering transformations were implemented in both Notebook 1 and Notebook 3 using different technologies.

| Aspect | Pandas / scikit-learn | PySpark |
|---|---|---|
| Data handling | Pandas DataFrame | PySpark DataFrame |
| Feature creation | Pandas operations | PySpark DataFrame APIs |
| Categorical encoding | scikit-learn `OneHotEncoder` | Spark `StringIndexer` + `OneHotEncoder` |
| Execution | Mainly local/in-memory | Distributed processing |
| Scalability | Suitable for smaller datasets | Suitable for large datasets |

The same feature-engineering logic was maintained while adapting the implementation to the Spark framework. The PySpark implementation demonstrates how the transformations can be performed using Spark DataFrame and ML APIs without rebuilding the complete machine-learning workflow.

### Task 8 — Why Spark Is Useful for Large Data

Spark becomes useful when datasets are too large to be efficiently processed on a single machine. It can distribute data and computations across multiple machines, allowing large datasets to be processed in parallel. This makes data processing and feature engineering more scalable and efficient than processing everything on a single machine.
