# Notebook 1 — Train & Save

This notebook contains the main machine-learning workflow for predicting customer churn.

The workflow starts with the raw Telco Customer Churn dataset and covers data exploration, evaluation strategy, baseline modelling, feature engineering, tree-based modelling, hyperparameter tuning, model comparison, final test evaluation, error analysis, visual analysis, and saving the final trained Pipeline.

The final objective is to select a suitable churn prediction model and save the complete preprocessing and modelling Pipeline so that it can later be loaded and used for prediction without retraining.

## Notebook Workflow

The workflow followed in this notebook is:

1. Understand and explore the dataset.
2. Create a train-validation-test evaluation strategy.
3. Build and evaluate a baseline Logistic Regression model.
4. Apply feature engineering and preprocessing.
5. Train tree-based models.
6. Tune the Random Forest model using cross-validation.
7. Compare model performance fairly.
8. Evaluate the selected model on the untouched test set.
9. Perform error and post-training analysis.
10. Save the complete final Pipeline for future predictions.

- **Accuracy** – Overall correct predictions.
- **Precision** – Correct positive predictions.

- **ROC-AUC** – Measures classification performance.
- **Confusion Matrix** – Shows TP, TN, FP, and FN.
- **Classification Report** – Summary of precision, recall, F1-score, and support.
- **ROC Curve** – Shows TPR vs FPR.
- **Precision-Recall Curve** – Shows precision vs recall.
- **AUC** – Calculates the area under a curve.'

# Task 2 — Create the Evaluation Strategy

The dataset is prepared for supervised classification by separating the input features from the churn target.

Columns that are identifiers, redundant information, target-related information, or potential sources of data leakage are removed before modelling.

The remaining data is divided into training, validation, and test sets using a stratified split so that the churn class distribution remains reasonably consistent across the datasets.

The validation set is used during model development and comparison, while the test set is kept untouched until the final model has been selected.

## Primary Evaluation Metric

The primary metric used for model comparison is **F1 Score**.

F1 Score is appropriate for churn prediction because it balances precision and recall. This is useful when identifying customers who may churn, since both missed churners and incorrect churn predictions are important.

ROC-AUC is also reported as a supporting metric to evaluate the model's ability to distinguish between churn and non-churn customers across classification thresholds.

## Task 2 Workflow

1. Remove irrelevant and leakage-prone columns.
2. Separate features `X` and target `y`.
3. Inspect the target distribution.
4. Split the data into training, validation, and test sets.
5. Use stratification to preserve class proportions.
6. Keep the test set untouched until final evaluation.
7. Use F1 Score as the primary model-selection metric.

###  Feature and Target Preparation

This code removes **irrelevant and leakage-prone columns**, separates the dataset into **features (`X`) and target (`y`)**, and converts churn labels into **0 (No)** and **1 (Yes)** for model training.


```python
drop_columns = [
    "CustomerID",
    "Count",
    "Country",
    "State",
    "Churn Label",
    "Churn Value",
    "Churn Score",
    "Churn Reason"
]

X = df.drop(columns=drop_columns)
y = df["Churn Label"].map({
    "No": 0,
    "Yes": 1
})

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

###  Target Distribution

This code displays the **number and percentage of customers in each target class** (0 = No Churn, 1 = Churn) to check the class distribution.


```python
print(y.value_counts())

print("\nTarget percentage:")
print(
    y.value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

###  Train, Validation & Test Split

This code splits the dataset into **training (64%)**, **validation (16%)**, and **testing (20%)** sets using `stratify` to maintain the same churn class distribution in each set.


```python
X_train_temp, X_test, y_train_temp, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

X_train, X_val, y_train, y_val = train_test_split(
    X_train_temp,
    y_train_temp,
    test_size=0.20,
    random_state=42,
    stratify=y_train_temp
)
```

```python
print("Training:", X_train.shape)
print("Validation:", X_val.shape)
print("Test:", X_test.shape)
```

### Target Distribution Check

This code checks the target class proportions in the training, validation, and test sets to ensure that the churn distribution remains consistent across all three datasets after stratified splitting.

```python
print("\nTrain target distribution:")
print(y_train.value_counts(normalize=True).round(3))

print("\nValidation target distribution:")
print(y_val.value_counts(normalize=True).round(3))

print("\nTest target distribution:")
print(y_test.value_counts(normalize=True).round(3))
```

# Task 3 — Build a Baseline Model

A baseline Logistic Regression model is trained before applying feature engineering.

The baseline establishes a reference level of performance against which more advanced modelling approaches can be compared.

The model uses the preprocessing pipeline defined earlier, including numerical imputation and scaling and categorical imputation and One-Hot Encoding.

The baseline is trained only on the training data and evaluated on the validation set.

## Task 3 Workflow

1. Create the baseline Logistic Regression pipeline.
2. Fit the model using the training data.
3. Generate validation predictions.
4. Generate churn probabilities.
5. Calculate Accuracy, Precision, Recall, F1 Score, and ROC-AUC.
6. Record the baseline performance for later comparison.

###  Feature Type Identification

This code separates the features into **numerical** and **categorical** columns based on their data types, which helps prepare them for appropriate preprocessing.


```python
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])
```

###  Numerical Preprocessing

This pipeline handles missing numerical values using the **median** and then **standardizes** the features using `StandardScaler`.


```python
numeric_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["str"]
).columns.tolist()

print("Numerical features:")
print(numeric_features)

print("\nCategorical features:")
print(categorical_features)
```

###  Categorical Preprocessing

This pipeline fills missing categorical values using the **most frequent value** and converts categorical features into **numerical form** using **One-Hot Encoding**.


```python
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

###  Combined Preprocessing

This code combines the **numerical and categorical preprocessing pipelines** using `ColumnTransformer` and applies the appropriate transformation to each feature type.


```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features), 
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

###  Baseline Model

This code creates a **Logistic Regression baseline model** by combining the preprocessing pipeline with the classifier. It provides a simple benchmark for evaluating model performance.


```python
baseline_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```

###  Model Training

`baseline_model.fit(X_train, y_train)` trains the **Logistic Regression model** using the training data after applying the defined preprocessing steps.


```python
baseline_model.fit(X_train, y_train)
```

###  Validation Predictions

This code generates **predicted churn classes** and **churn probabilities** for the validation dataset to evaluate the model's performance.


```python
y_val_pred = baseline_model.predict(X_val)

y_val_prob = baseline_model.predict_proba(X_val)[:, 1]
```

###  Baseline Model Evaluation

This code calculates and displays the **Accuracy, Precision, Recall, F1 Score, and ROC-AUC** of the Logistic Regression model on the validation data.


```python
baseline_accuracy = accuracy_score(y_val, y_val_pred)
baseline_precision = precision_score(y_val, y_val_pred)
baseline_recall = recall_score(y_val, y_val_pred)
baseline_f1 = f1_score(y_val, y_val_pred)
baseline_roc_auc = roc_auc_score(y_val, y_val_prob)

print("Baseline Logistic Regression")
print("--------------------------------")
print("Accuracy :", round(baseline_accuracy, 4))
print("Precision:", round(baseline_precision, 4))
print("Recall   :", round(baseline_recall, 4))
print("F1 Score :", round(baseline_f1, 4))
print("ROC-AUC  :", round(baseline_roc_auc, 4))
```

###  Baseline Model Evaluation

This code calculates and displays the **Accuracy, Precision, Recall, F1 Score, and ROC-AUC** of the Logistic Regression model on the validation data.


```python
print(classification_report(
    y_val,
    y_val_pred,
    target_names=["No Churn", "Churn"]
))
```

# Task 4 — Feature Engineering

Feature engineering creates meaningful features from raw customer data to improve churn prediction.

* **Tenure Years** – Converts tenure from months to years.
* **Average Monthly Charge** – Calculates average charges based on tenure.
* **Service Count** – Counts subscribed services.
* **Charge-Tenure Interaction** – Combines monthly charges and tenure.

### Categorical Encoding

Categorical features are converted using **One-Hot Encoding** without creating artificial ordering.

### Scaling

Numerical features are **median-imputed and standardized** using `StandardScaler`. Scaling mainly helps Logistic Regression.

### Missing Values

* Numerical → **Median**
* Categorical → **Most Frequent**

### Leakage Check

Feature engineering does **not use the churn target or future information**, helping prevent data leakage.


###  Feature Engineering

This code creates **new meaningful features** such as tenure in years, average monthly charges, service count, and charge-tenure interaction to improve the model's ability to predict customer churn.


```python
from feature_engineering import ChurnFeatureEngineer
```

###  Apply Feature Engineering

This code applies the **feature engineering class** to the training data and compares the **original number of features** with the **new engineered feature set**.


```python
feature_engineer = ChurnFeatureEngineer()

X_train_engineered = feature_engineer.fit_transform(X_train)

print("Original shape:", X_train.shape)
print("Engineered shape:", X_train_engineered.shape)
```

###  Engineered Features Preview

This code displays the **first few rows** of the selected original and newly engineered features to verify that the feature engineering was applied correctly.


```python
X_train_engineered[
    [
        "Tenure Months",
        "Tenure Years",
        "Total Charges",
        "Average Monthly Charge",
        "Service Count",
        "Monthly Charges",
        "Charge Tenure Interaction"
    ]
].head()
```

### Domain-Driven Features

The engineered features are domain-driven because they represent customer tenure, spending behaviour, service usage, and the interaction between charges and tenure. No date-derived feature was created because the dataset does not contain a suitable date variable.

###  Missing Value Check

This code checks for **missing values in the newly engineered features** to ensure they are ready for further preprocessing and model training.


```python
X_train_engineered[
    [
        "Tenure Years",
        "Average Monthly Charge",
        "Service Count",
        "Charge Tenure Interaction"
    ]
].isnull().sum()
```

###  Engineered Feature Types

This code separates the **engineered dataset into numerical and categorical features** based on their data types for appropriate preprocessing.


```python
numeric_features_fe = X_train_engineered.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features_fe = X_train_engineered.select_dtypes(
    include=["str"]
).columns.tolist()

print("Engineered numerical features:")
print(numeric_features_fe)

print("\nEngineered categorical features:")
print(categorical_features_fe)
```

###  Preprocessing Engineered Features

This code creates separate preprocessing pipelines for **numerical and categorical features**, handling missing values and applying scaling or one-hot encoding.


```python
numeric_transformer_fe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_transformer_fe = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])
```

###  Combined Preprocessing

This code combines the **numerical and categorical preprocessing pipelines** using `ColumnTransformer` to apply the correct preprocessing to each feature type.


```python
preprocessor_fe = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer_fe, numeric_features_fe),
        ("cat", categorical_transformer_fe, categorical_features_fe)
    ]
)
```

###  Engineered Model Pipeline

This code creates a **Logistic Regression pipeline** that performs **feature engineering, preprocessing, and model training** sequentially.


```python
engineered_lr_pipeline = Pipeline(steps=[
    ("feature_engineering", ChurnFeatureEngineer()),
    ("preprocessor", preprocessor_fe),
    ("model", LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```

###  Engineered Model Training

`engineered_lr_pipeline.fit(X_train, y_train)` trains the **engineered Logistic Regression model** on the training data, including feature engineering and preprocessing.


```python
engineered_lr_pipeline.fit(X_train, y_train)
```

###  Engineered Model Predictions

This code generates **churn predictions and churn probabilities** for the validation dataset using the engineered Logistic Regression pipeline.


```python
engineered_pred = engineered_lr_pipeline.predict(X_val)
engineered_prob = engineered_lr_pipeline.predict_proba(X_val)[:, 1]
```

###  Engineered Model Evaluation

This code calculates and displays the **Accuracy, Precision, Recall, F1 Score, and ROC-AUC** of the engineered Logistic Regression model on the validation data.


```python
engineered_accuracy = accuracy_score(y_val, engineered_pred)
engineered_precision = precision_score(y_val, engineered_pred)
engineered_recall = recall_score(y_val, engineered_pred)
engineered_f1 = f1_score(y_val, engineered_pred)
engineered_roc_auc = roc_auc_score(y_val, engineered_prob)
print("Engineered Logistic Regression")
print("--------------------------------")
print("Accuracy :", round(engineered_accuracy, 4))
print("Precision:", round(engineered_precision, 4))
print("Recall   :", round(engineered_recall, 4))
print("F1 Score :", round(engineered_f1, 4))
print("ROC-AUC  :", round(engineered_roc_auc, 4))
```

# Task 6 — Train Tree-Based Models

Tree-based models are used to capture nonlinear relationships and interactions between customer features.

A Random Forest classifier is used as a bagging-based model, while XGBoost is used as a boosting-based model. Both models are trained using the same training data and evaluated on the same validation set for a fair comparison.

## XGBoost Model

XGBoost is a gradient-boosting algorithm that builds trees sequentially, with each new tree focusing on correcting errors made by previous trees.

###  XGBoost Model

This code creates an **XGBoost classification pipeline** that performs feature engineering, preprocessing, and churn prediction using the **XGBClassifier**.


```python
xgb_model = Pipeline(steps=[
    ("feature_engineering", ChurnFeatureEngineer()),
    ("preprocessor", preprocessor),
    ("model", XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric="logloss"
    ))
])
```

###  Train XGBoost Model

This code trains the **XGBoost model** on the training data after applying feature engineering and preprocessing.


```python
xgb_model.fit(X_train, y_train)
```

###  XGBoost Validation Performance

This code generates predictions on the **validation set** and evaluates XGBoost using **Accuracy, Precision, Recall, F1 Score, and ROC-AUC**.


```python
xgb_val_pred = xgb_model.predict(X_val)
xgb_val_prob = xgb_model.predict_proba(X_val)[:, 1]

xgb_accuracy = accuracy_score(y_val, xgb_val_pred)
xgb_precision = precision_score(y_val, xgb_val_pred)
xgb_recall = recall_score(y_val, xgb_val_pred)
xgb_f1 = f1_score(y_val, xgb_val_pred)
xgb_roc_auc = roc_auc_score(y_val, xgb_val_prob)

print("XGBoost Validation Performance")
print("Accuracy :", round(xgb_accuracy, 4))
print("Precision:", round(xgb_precision, 4))
print("Recall   :", round(xgb_recall, 4))
print("F1 Score :", round(xgb_f1, 4))
print("ROC-AUC  :", round(xgb_roc_auc, 4))
```

## Feature Importance — XGBoost

Feature importance is examined to identify which customer features contribute most to the XGBoost predictions. This provides insight into which variables are most useful for distinguishing churned and non-churned customers.

```python
xgb_feature_names = xgb_model["preprocessor"].get_feature_names_out()
xgb_importances = xgb_model["model"].feature_importances_

xgb_feature_importance = pd.DataFrame({
    "Feature": xgb_feature_names,
    "Importance": xgb_importances
}).sort_values(
    "Importance",
    ascending=False
)

xgb_feature_importance.head(10)
```

### Top 10 XGBoost Feature Importances

The top 10 feature importance scores are visualized to identify the features that contribute most to the XGBoost model's churn predictions.

```python
top10_xgb = xgb_feature_importance.head(10).sort_values("Importance")

plt.figure(figsize=(8, 5))

plt.barh(
    top10_xgb["Feature"],
    top10_xgb["Importance"]
)

plt.title("Top 10 Feature Importances - XGBoost")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()
```

###  Random Forest Model

This code creates a **Random Forest pipeline** that performs feature engineering, preprocessing, and classification. `class_weight="balanced"` helps handle **class imbalance** in the churn dataset.


```python
rf_pipeline = Pipeline(steps=[
    ("feature_engineering", ChurnFeatureEngineer()),
    ("preprocessor", preprocessor_fe),
    ("model", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1,
        class_weight="balanced"
    ))
])
```

###  Train Random Forest Model

`rf_pipeline.fit(X_train, y_train)` trains the **Random Forest model** on the training data after applying feature engineering and preprocessing.


```python
rf_pipeline.fit(X_train, y_train)
```

###  Random Forest Evaluation

This code generates predictions and calculates **Accuracy, Precision, Recall, F1 Score, and ROC-AUC** to evaluate the Random Forest model on the validation data.


```python
rf_val_pred = rf_pipeline.predict(X_val)
rf_val_prob = rf_pipeline.predict_proba(X_val)[:, 1]

rf_accuracy = accuracy_score(y_val, rf_val_pred)
rf_precision = precision_score(y_val, rf_val_pred)
rf_recall = recall_score(y_val, rf_val_pred)
rf_f1 = f1_score(y_val, rf_val_pred)
rf_roc_auc = roc_auc_score(y_val, rf_val_prob)

print("Random Forest")
print("----------------------------")
print("Accuracy :", round(rf_accuracy, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("F1 Score :", round(rf_f1, 4))
print("ROC-AUC  :", round(rf_roc_auc, 4))
```

### Model Complexity Analysis

Different values of `max_depth` are tested to examine how tree complexity affects training and validation performance.

A large difference between training and validation F1 may indicate overfitting, while consistently low scores may indicate underfitting. The experiment helps understand the trade-off between model complexity and generalization.

```python
depth_results = []

for depth in [3, 5, 10, None]:

    rf_experiment = Pipeline(steps=[
        ("feature_engineering", ChurnFeatureEngineer()),
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=200,
            max_depth=depth,
            random_state=42,
            class_weight="balanced"
        ))
    ])

    rf_experiment.fit(X_train, y_train)

    train_pred = rf_experiment.predict(X_train)
    val_pred = rf_experiment.predict(X_val)

    depth_results.append({
        "Max Depth": depth,
        "Train F1": f1_score(y_train, train_pred),
        "Validation F1": f1_score(y_val, val_pred)
    })

depth_results_df = pd.DataFrame(depth_results)

depth_results_df
```

## Bagging vs Boosting

Random Forest uses bagging, where multiple decision trees are trained independently on different samples and their predictions are combined. This mainly helps reduce variance and improve stability.

XGBoost uses boosting, where trees are built sequentially and each new tree focuses on correcting errors from previous trees. This can produce strong predictive performance but may require careful tuning to avoid overfitting.

In this project, Random Forest represents the bagging approach and XGBoost represents the boosting approach.

# Task 10 — Error Analysis

Error analysis examines the incorrect predictions made by the final tuned Random Forest model on the unseen test set.

The analysis identifies false positives and false negatives and examines how the model's prediction errors vary across customer segments.

Contract type is used as a meaningful customer segment to investigate whether prediction errors are concentrated in particular groups.

This analysis can help identify customer groups where the model may require further improvement.

###  Error Analysis

This code compares the **actual and predicted churn values** to identify incorrect predictions and categorizes them as **False Positive, False Negative, or Correct**.


```python
error_analysis = X_test.copy()

error_analysis["Actual"] = y_test
error_analysis["Predicted"] = test_pred

error_analysis["Error"] = (
    error_analysis["Actual"] != error_analysis["Predicted"]
)

error_analysis["Error Type"] = np.select(
    [
        (error_analysis["Actual"] == 0) & (error_analysis["Predicted"] == 1),
        (error_analysis["Actual"] == 1) & (error_analysis["Predicted"] == 0)
    ],
    [
        "False Positive",
        "False Negative"
    ],
    default="Correct"
)

print("Error analysis completed.")
print(
    error_analysis["Error Type"].value_counts()
)
```

###  Error Rate by Contract Type

This code calculates the **prediction error rate for each contract type** to identify which contract groups are more difficult for the model to predict correctly.


```python
contract_error = (
    error_analysis
    .groupby("Contract")["Error"]
    .mean()
    .mul(100)
    .sort_values(ascending=False)
)

print("Error Rate by Contract Type:")
print(contract_error.round(2))
```

###  Final Model Error Rate by Contract Type

This code creates a **bar chart** showing the model's **prediction error rate for each contract type**, making it easier to compare where the model makes more errors.


```python
plt.figure(figsize=(8, 5))

plt.bar(
    contract_error.index,
    contract_error.values
)

plt.title("Final Model Error Rate by Contract Type")
plt.xlabel("Contract Type")
plt.ylabel("Error Rate (%)")

plt.tight_layout()
plt.show()
```

### Error Analysis Insight

The error-rate comparison shows how the final model's incorrect predictions vary across contract types.

Contract groups with higher error rates may contain customer patterns that are more difficult for the model to distinguish. These groups could be investigated further through additional feature engineering, threshold adjustment, or alternative modelling approaches.

### Suggested Improvement

One improvement I would investigate next is **adjusting the classification threshold**.

The model has more false positives than false negatives. Testing different probability thresholds could help find a better balance between precision and recall and potentially improve the F1 Score.

2. **Payment Method:** Customers using electronic check have the highest churn rate at 45.29%. In comparison, customers using automatic bank transfer and automatic credit card payments have much lower churn rates.

3. **Contract Type:** Month-to-month customers show considerably higher churn compared with customers on one-year or two-year contracts.

4. **Internet Service:** Fiber optic customers show higher churn than DSL and customers without internet service.

5. **Customer Profile:** Senior citizens have a higher churn rate than non-senior customers, while customers without partners show higher churn than customers with partners.

6. **Model Findings:** The Random Forest models achieved better churn detection than the Logistic Regression models. The tuned Random Forest achieved the strongest validation ROC-AUC and Recall among the evaluated models.
