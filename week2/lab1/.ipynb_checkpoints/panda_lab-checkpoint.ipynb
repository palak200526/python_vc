## Import Pandas and Load the Excel Dataset

The `read_excel()` function is used to read data from the Excel file, and the data is stored in the variable `df` for further analysis and manipulation.

```python
import pandas as pd
df= pd.read_excel("business.xlsx")
```

## Converting Data to Numeric and Filtering Values

The `Footnotes` column is converted into numeric values using `pd.to_numeric()`.

The `errors="coerce"` parameter converts values that cannot be converted into numbers to `NaN`.

After conversion, the condition `df["Footnotes"] > 5` checks which values in the `Footnotes` column are greater than `5`.

The `print()` statement displays the resulting Boolean values (`True` or `False`).

```python
df["Footnotes"]= pd.to_numeric(df["Footnotes"], errors="coerce")
print(df["Footnotes"]>5)
```

## Load the Student Dataset

In this section, the `student.csv` file is loaded into a Pandas DataFrame using the `read_csv()` function.

The dataset is stored in the variable `df`, which will be used for further data analysis and manipulation.

```python
df=pd.read_csv("student.csv")
```

## Display the Dataset

The DataFrame `df` is displayed to view the complete dataset and inspect the available rows and columns.

```python
df
```

## Filtering Rows Using Multiple Conditions

In this section, rows are filtered based on two conditions.

We select students whose `study_time_hours` is greater than 4 **and** whose `attendance_percent` is greater than 45.

The `&` operator is used to apply both conditions simultaneously.

```python
df[(df["study_time_hours"]>4) & (df["attendance_percent"]>45 )]
```

## Creating a New Calculated Column

In this section, a new column named `Average Grade` is created by calculating the average of the `previous_grade` and `final_exam_score` columns.

The calculated value is stored in the new `Average Grade` column for each student.

```python
df["Average Grade"]= (df["previous_grade"] + df["final_exam_score"])/2
```

## Display the Average Grade Column

The `Average Grade` column is displayed to view the calculated average grade of each student.

```python
df["Average Grade"]
```

## Display the Updated DataFrame

The complete DataFrame is displayed to view the dataset along with the newly created `Average Grade` column.

```python
df
```

## GroupBy and Aggregation

In this section, the data is grouped based on the `part_time_job` column.

For each group, the average study time is calculated using `mean`, and the number of students in each group is calculated using `count`.

The result is stored in the `result` variable and then displayed using `print()`.

```python
result = df.groupby("part_time_job")["study_time_hours"].agg(["mean", "count"])

print(result)
```

## Creating a DataFrame

In this section, a new DataFrame named `df1` is created using a dictionary.

The DataFrame contains two columns: `student_id` and `name`, with three student records.

The `print()` function is used to display the created DataFrame.

```python
df1 = pd.DataFrame({
    "student_id": [1, 2, 3],
    "name": ["James", "John", "Mary"]
})

print(df1)
```

## Creating a Second DataFrame

In this section, a second DataFrame named `df2` is created using a dictionary.

The DataFrame contains two columns: `student_id` and `marks`. The `student_id` column will be used as the common column when merging `df1` and `df2`.

The `print()` function is used to display the created DataFrame.

```python
df2 = pd.DataFrame({
    "student_id": [1, 2, 3],
    "marks": [85, 90, 78]
})
print(df2)
```

## Merging Two DataFrames

In this section, the two DataFrames `df1` and `df2` are merged using the common `student_id` column.

The `pd.merge()` function combines the information from both DataFrames based on the matching `student_id` values.

The merged DataFrame is stored in `result` and displayed using `print()`.

```python
result = pd.merge(df1, df2, on="student_id")

print(result)
```

## Checking Row Counts After Merging

In this section, the number of rows in `df1`, `df2`, and the merged DataFrame `result` is checked.

The `len()` function is used to count the rows in each DataFrame. This helps verify the result of the merge and compare the number of records before and after merging.

```python
print("df1 rows:", len(df1))
print("df2 rows:", len(df2))
print("Merged rows:", len(result))
```
