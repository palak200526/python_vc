## Check Missing Values

The `isnull()` function is used to identify missing values in the dataset. The `sum()` function then counts the number of missing values in each column.


```python
missing_count = df.isnull().sum()
print(missing_count)
```

## Calculate Missing Value Percentage

The missing values in each column are calculated as a percentage of the total number of rows. This helps us understand how much of the dataset is affected by missing data.

### Explanation

- `df.isnull().sum()` counts the missing values in each column.
- `len(df)` gives the total number of rows in the dataset.
- Dividing the missing count by the total rows gives the proportion of missing values.
- Multiplying by `100` converts the proportion into a percentage.
- The resulting `missing_percentage` shows the percentage of missing values in each column.

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100
print(missing_percentage)
```

## Create Missing Values Summary

A summary DataFrame is created to combine the number of missing values and their corresponding percentages for each column. This makes it easier to review the missing data in one place.

### Explanation

- `pd.DataFrame()` creates a new DataFrame.
- `"Missing Count"` stores the total number of missing values in each column.
- `"Missing Percentage"` stores the percentage of missing values in each column.
- `missing_summary` contains both measures together for easier analysis.
- `print(missing_summary)` displays the complete missing-value summary.

```python
missing_summary= pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})
print(missing_summary)
```

## Check UNKNOWN and ERROR Values

The dataset may contain invalid text values such as `UNKNOWN` and `ERROR`. This step checks the values present in each column and displays their frequencies.

### Explanation

## Replace Invalid Values

The values `UNKNOWN` and `ERROR` are treated as invalid or missing data. They are replaced with Pandas' `NA` value so that they can be handled like other missing values during data cleaning and analysis.

### Explanation

- `df.replace()` searches the entire DataFrame for specific values.
- `["UNKNOWN", "ERROR"]` specifies the invalid values to be replaced.
- `pd.NA` represents a missing value in Pandas.
- After this step, `UNKNOWN` and `ERROR` will be treated as missing values and can be handled using missing-value techniques.

```python
df=df.replace(["UNKNOWN", "ERROR"], pd.NA)
```

```python
for column in df.columns:
    print("\n", column)
    print(df[column].value_counts(dropna=False).head(15))
```

## Handle Missing Values in the Item Column

Missing values in the `Item` column are filled using the **mode**, which is the most frequently occurring value in that column.

### Explanation

- `df["Item"]` selects the `Item` column.
- `.mode()` finds the most frequently occurring value in the column.
- `[0]` selects the first mode value.
- `.fillna()` replaces missing values with the selected mode.
- This approach is suitable for a categorical column such as `Item`.

```python
df["Item"]=df["Item"].fillna(df["Item"].mode()[0])
```

## Handle Missing Values in the Quantity Column

The `Quantity` column is converted into a numeric data type, and any invalid or missing values are replaced with the median of the column.

### Explanation

- `pd.to_numeric()` converts the values in the `Quantity` column into numeric values.
- `errors="coerce"` converts values that cannot be converted into numbers into missing values (`NaN`).
- `.median()` calculates the middle value of the valid quantities.
- `.fillna()` replaces missing values with the calculated median.
- Using the median helps fill missing numerical values while reducing the effect of extreme values.

```python
df["Quantity"]= pd.to_numeric(df["Quantity"], errors="coerce")
df["Quantity"]=df["Quantity"].fillna(df["Quantity"].median())
```

## Handle Missing Values in the Price Per Unit Column

The `Price Per Unit` column is converted into a numeric data type, and invalid or missing values are replaced with the median of the column.

### Explanation

- `pd.to_numeric()` converts the values in the `Price Per Unit` column into numeric values.
- `errors="coerce"` converts invalid values into missing values (`NaN`).
- `.median()` calculates the median price from the valid numeric values.
- `.fillna()` replaces missing values with the calculated median.
- Using the median helps maintain a reasonable value while reducing the influence of unusually high or low prices.

```python
df["Price Per Unit"]=pd.to_numeric(df["Price Per Unit"], errors="coerce")
df["Price Per Unit"]=df["Price Per Unit"].fillna(df["Price Per Unit"].median())
```

## Handle Missing Values in the Total Spent Column

The `Total Spent` column is converted into a numeric data type, and invalid or missing values are replaced with the median of the column.

## Detect Outliers in the Quantity Column

The **Interquartile Range (IQR) method** is used to identify unusual or extreme values in the `Quantity` column.

### Explanation

- `Q1` represents the 25th percentile of the `Quantity` values.
- `Q3` represents the 75th percentile of the `Quantity` values.
- `IQR` is calculated as the difference between Q3 and Q1.
- `lower_bound` defines the minimum expected value using `Q1 - 1.5 × IQR`.
- `upper_bound` defines the maximum expected value using `Q3 + 1.5 × IQR`.
- Values below the lower bound or above the upper bound are considered **outliers**.
- The `outliers` DataFrame contains all rows where the `Quantity` value falls outside these boundaries.
- `print(outliers)` displays the detected outlier records.

```python
Q1=df["Quantity"].quantile(0.25)
Q3=df["Quantity"].quantile(0.75)
IQR= Q3 - Q1
lower_bound= Q1 - 1.5*IQR
upper_bound= Q3 + 1.5*IQR
outliers= df[
    (df["Quantity"] < lower_bound) |
    (df["Quantity"] > upper_bound)
]
print(outliers)
```

```python
Q1= df["Price Per Unit"].quantile(0.25)
Q3= df["Price Per Unit"].quantile(0.75)
IQR= Q3 - Q1
lower_bound= Q1 - 1.5*IQR 
upper_bound= Q3 + 1.5*IQR
outliers= df[
    (df["Price Per Unit"] < lower_bound) | 
    (df["Price Per Unit"] > upper_bound)
]
print(outliers)
```

```python
Q1= df["Total Spent"].quantile(0.25)
Q3= df["Total Spent"].quantile(0.75)
IQR= Q3 - Q1
lower_bound= Q1 - 1.5*IQR
upper_bound= Q3 + 1.5*IQR
outliers= df[
    (df["Total Spent"] < lower_bound) | 
    (df["Total Spent"] > upper_bound)
]
print(outliers)
```
