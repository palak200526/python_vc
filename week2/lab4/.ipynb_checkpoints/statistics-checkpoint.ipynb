## Load the Dataset

The following code imports the Pandas library and loads the `marketAnalysis.csv` dataset into a DataFrame named `df`.


````python
import pandas as pd
df= pd.read_csv("marketAnalysis.csv")
````

### DataFrame

The variable `df` represents the Pandas DataFrame containing the data loaded from the `marketAnalysis.csv` file. It is used to access, analyze, and manipulate the dataset.

````python
df
````

### Select the Sales Column

The following code selects the `Sales` column from the DataFrame and stores it in a variable named `sales`.


````python
sales= df["Sales"]
````

### Mean

The mean represents the average value of the `Sales` column.

**Formula:**

$$
\text{Mean} = \frac{\sum x_i}{n}
$$

where:

- \(x_i\) = each sales value
- \(n\) = total number of sales values


````python
mean= sales.mean()
print("Mean: ", mean)
````

### Median

The median represents the middle value of the `Sales` column when all values are arranged in ascending order.

**Formula:**

$$
\text{Median} =
\begin{cases}
x_{\frac{n+1}{2}}, & \text{if } n \text{ is odd} \\
\frac{x_{\frac{n}{2}} + x_{\frac{n}{2}+1}}{2}, & \text{if } n \text{ is even}
\end{cases}
$$

where:

- $x$ = ordered sales values
- $n$ = total number of sales values

````python
median= sales.median()
print("Median: ",median) 
````

### Standard Deviation

Standard deviation measures how much the `Sales` values vary or spread out from the mean.

**Formula:**

$$
\sigma = \sqrt{\frac{\sum (x_i-\mu)^2}{N}}
$$

where:

- $x_i$ = each sales value
- $\mu$ = mean of the sales values
- $N$ = total number of sales values
- $\sigma$ = standard deviation


````python
std= sales.std()
print("Standard Deviation: ", std)
````

### Percentiles

Percentiles divide the data into parts and show the value below which a certain percentage of the observations fall.

**Formula:**

$$
P_k = \text{value below which } k\% \text{ of the data falls}
$$

The percentiles used here are:

- **25th percentile (Q1)**: 25% of the sales values are below this value.
- **50th percentile (Q2)**: 50% of the sales values are below this value. This is also the **median**.
- **75th percentile (Q3)**: 75% of the sales values are below this value.
- **90th percentile**: 90% of the sales values are below this value.
- **95th percentile**: 95% of the sales values are below this value.


````python
percentiles= sales.quantile([0.25, 0.50, 0.75, 0.90, 0.95])
print("Percentiles: \n", percentiles)
````

## Sales Distribution Histogram

The histogram is used to visualize the **distribution of sales values**.

## Explanation

* `plt.figure(figsize=(8,5))` sets the figure size to 8 × 5 inches.
* `plt.hist(sales, bins=30, edgecolor="black")` creates a histogram using the `sales` data.

  * `bins=30` divides the sales values into 30 intervals.
  * `edgecolor="black"` adds borders around the bars for better visibility.
* `plt.xlabel("Sales")` labels the x-axis as **Sales**.

## Sales Skewness

Skewness measures the **asymmetry of the sales distribution**.

## Explanation

* `sales.skew()` calculates the skewness of the `sales` data.
* The calculated value is stored in the variable `skewness`.
* `print("Skewness: ", skewness)` displays the skewness value.

## Interpretation

* **Skewness ≈ 0** → Sales distribution is approximately symmetric.
* **Skewness > 0** → Sales distribution is **right-skewed** (positively skewed).
* **Skewness < 0** → Sales distribution is **left-skewed** (negatively skewed).

The skewness value helps determine the **direction and degree of asymmetry** in the sales data.


````python
skewness= sales.skew()
print("Skewness: ", skewness)
````

## Outlier Detection Using IQR

The **Interquartile Range (IQR) method** is used to identify outliers in the sales data.

### Code Explanation

* `sales.quantile(0.25)` calculates the **first quartile (Q1)**, which represents the 25th percentile.
* `sales.quantile(0.75)` calculates the **third quartile (Q3)**, which represents the 75th percentile.
* `IQR = Q3 - Q1` calculates the **Interquartile Range**.
* `lower_bound = Q1 - 1.5 * IQR` calculates the lower limit for detecting outliers.
* `upper_bound = Q3 + 1.5 * IQR` calculates the upper limit for detecting outliers.
* `sales[(sales < lower_bound) | (sales > upper_bound)]` selects values that are below the lower bound or above the upper bound.


````python
Q1= sales.quantile(0.25)
Q3= sales.quantile(0.75)
IQR= Q3 - Q1
lower_bound= Q1 - 1.5*IQR
upper_bound= Q3 + 1.5*IQR
outliers= sales[(sales < lower_bound) | (sales > upper_bound)]
print("Q1: ", Q1)
print("Q3: ", Q3)
print("IQR: ",IQR)
print("Lower Bound: ", lower_bound)
print("Upper Bound: ", upper_bound)
print("Outliers: \n", outliers)
````

## Correlation Matrix

A correlation matrix shows the **strength and direction of the relationship** between numerical variables.

* Values range from **-1 to +1**.
* **Positive value** → variables increase together.
* **Negative value** → one increases while the other decreases.
* **Near 0** → weak or no linear relationship.

It helps identify **strong relationships between numerical variables**.


````python
corr_matrix = df.corr(numeric_only=True)

print(corr_matrix)
````

## Correlation Pairs

This code converts the correlation matrix into a **single series of correlation pairs**.

```python
corr = df.corr(numeric_only=True)
corr_pairs = corr.unstack()
print(corr_pairs)
```

* `df.corr()` calculates the correlation matrix.
* `unstack()` converts the matrix into a **Series of variable pairs and their correlation values**.
* `print()` displays all correlation pairs.

This makes it easier to find the **strongest relationships** between numerical variables.


````python
corr = df.corr(numeric_only= True)
corr_pairs = corr.unstack()
print(corr_pairs)
````

## Removing Self-Correlations

This code removes correlations where a variable is compared with itself.


* `get_level_values(0)` gets the first variable name.
* `get_level_values(1)` gets the second variable name.
* `!=` keeps only pairs where the two variables are **different**.

This removes values like `Sales-Sales = 1.0` and keeps only **relationships between different variables**.


````python
corr_pairs= corr_pairs[
    corr_pairs.index.get_level_values(0) !=
    corr_pairs.index.get_level_values(1)
]
print(corr_pairs)
````

## Finding the Strongest Correlation Pair

This code finds the pair of numerical variables with the **strongest correlation**.

### Explanation

* `abs()` considers the **absolute value** of correlations, so both strong positive and negative correlations are considered.
* `sort_values(ascending=False)` sorts the correlations from strongest to weakest.
* `.index[0]` selects the pair with the **strongest correlation**.
* `corr.loc[...]` retrieves the actual correlation value, including its positive or negative sign.

````python
strongest_pair = corr_pairs.abs().sort_values(ascending= False).index[0]
print(strongest_pair)
print("Correlation :", corr.loc[strongest_pair[0], strongest_pair[1]])
````

## Independent T-Test

An **independent t-test** is used to compare the mean sales of two independent groups: **Male** and **Female** customers.

### Interpretation

* `males_sales` and `females_sales` contain the sales values for each group.
* `.dropna()` removes missing sales values.
* `ttest_ind()` performs an independent t-test.
* `equal_var=False` performs **Welch's t-test**, which does not assume equal variance between the two groups.
* **T-statistic** measures the difference between the group means relative to the variation in the data.
* **P-value** helps determine whether the observed difference is statistically significant.

If the **p-value < 0.05**, we reject the null hypothesis and conclude that there is a statistically significant difference in mean sales between the two groups.

If the **p-value ≥ 0.05**, we fail to reject the null hypothesis and do not have sufficient evidence to conclude that the mean sales are different.


````python
from scipy.stats import ttest_ind
males_sales= df[df["Gender"] == "Male"]["Sales"].dropna()
females_sales= df[df["Gender"] == "Female"]["Sales"].dropna()
# print("Males Sales: ", males_sales)
# print("Female Sales: ", females_sales)
t_stat, p_value = ttest_ind(males_sales, females_sales, equal_var=False)
print("T-Statistic: ", t_stat)
print("P-Value: ", p_value)
````

### Hypothesis Test Decision

The following condition is used to make a decision based on the **p-value**.

### Interpretation

* If **p-value < 0.05**, the result is statistically significant, so we **reject the null hypothesis**.
* If **p-value ≥ 0.05**, we **do not reject the null hypothesis**.

Here, the significance level is **0.05 (5%)**.


````python
if p_value < 0.05:
    print("Reject the null hypothesis")
else:
    print("Do not reject the null hypothesis")
````
