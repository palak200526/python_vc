## Importing Libraries and Loading the Dataset

The required Python libraries are imported and the `marketAnalysis.csv` dataset is loaded into a Pandas DataFrame.

* `pandas` is used for **data manipulation and analysis**.
* `matplotlib` is used for **creating visualizations**.
* `numpy` is used for **numerical operations**.
* `seaborn` is used for **statistical visualizations**.
* `pd.read_csv()` loads the CSV file into the DataFrame `df`.


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
df= pd.read_csv("marketAnalysis.csv")
```

## Displaying the DataFrame

The `df` command displays the **entire DataFrame**, allowing us to view the rows and columns of the dataset.


```python
df
```

## Distribution of Quantity Purchased

A histogram is used to visualize the **distribution of the quantity purchased**.

* `plt.figure(figsize=(8, 5))` sets the figure size.
* `sns.histplot()` creates the histogram for the `Quantity` column.
* `bins=10` divides the data into 10 intervals.
* The x-axis represents **Quantity**.
* The y-axis represents **Frequency**, or the number of observations.
* The histogram helps understand how frequently different quantities are purchased.


```python
plt.figure(figsize=(8, 5))

sns.histplot(df["Quantity"], bins=10)

plt.title("Distribution of Quantity Purchased")
plt.xlabel("Quantity")
plt.ylabel("Frequency")

plt.show()
```

## Boxplot of Quantity Purchased

A boxplot is used to visualize the **spread, median, and potential outliers** in the `Quantity` column.

* `sns.boxplot()` creates a boxplot for the `Quantity` column.
* The boxplot shows the **median, quartiles, and possible outliers**.
* The x-axis represents the **quantity purchased**.
* Values outside the whiskers may indicate **potential outliers**.


```python
plt.figure(figsize=(8, 4))

sns.boxplot(x=df["Quantity"])

plt.title("Boxplot of Quantity Purchased")
plt.xlabel("Quantity")

plt.show()
```

## Unit Price vs Quantity

A scatter plot is used to visualize the **relationship between Unit Price and Quantity**.

* `sns.scatterplot()` creates a scatter plot.
* `x="Unit price"` places **Unit Price** on the x-axis.
* `y="Quantity"` places **Quantity** on the y-axis.
* Each point represents an observation in the dataset.
* The plot helps identify whether there is a **relationship or pattern** between unit price and quantity purchased.


```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Unit price",
    y="Quantity"
)

plt.title("Unit Price vs Quantity")
plt.xlabel("Unit Price")
plt.ylabel("Quantity")

plt.show()
```

## Correlation Heatmap

A correlation heatmap is used to visualize the **relationships between numerical variables** in the dataset.

* `select_dtypes()` selects only numerical columns.
* `.corr()` calculates the correlation between numerical variables.
* `sns.heatmap()` displays the correlations visually.
* `annot=True` shows correlation values inside the cells.
* `fmt=".2f"` displays values up to 2 decimal places.
* Values close to **+1** indicate a strong positive relationship.
* Values close to **-1** indicate a strong negative relationship.
* Values close to **0** indicate a weak or no linear relationship.


```python
numeric_df = df.select_dtypes(include="number")

correlation = numeric_df.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```
