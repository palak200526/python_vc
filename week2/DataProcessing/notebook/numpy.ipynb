## Creating a NumPy Array

This creates a **1D NumPy array** containing numerical values.

- **`np.array()`** → Converts the values into a NumPy array.
- NumPy arrays are useful for **fast numerical calculations**.

### Key Idea

A NumPy array stores numerical data in an **efficient and structured format**.

```python
import numpy as np
import pandas as pd
data = np.array([10, 20, 30, 40, 50])

print(data)
```

## Calculating IQR

This calculates the **Interquartile Range (IQR)** of the alcohol data.

- **Q1** → 25th percentile.
- **Q3** → 75th percentile.
- **IQR** → Difference between Q3 and Q1.
- IQR represents the spread of the **middle 50% of the data**.

### Key Idea

IQR is useful for understanding **data spread and detecting outliers**.

```python
Q1 = np.percentile(alcohol, 25)
Q3 = np.percentile(alcohol, 75)
IQR = Q3 - Q1

print(IQR)
```

## Detecting Outliers Using IQR

This uses the **IQR method** to detect outliers in the alcohol data.

- **Q1** → 25th percentile.
- **Q3** → 75th percentile.
- **IQR** → `Q3 - Q1`.
- **Lower Bound** → `Q1 - 1.5 × IQR`.
- **Upper Bound** → `Q3 + 1.5 × IQR`.
- Values outside these bounds are identified as **outliers**.

### Key Idea

The IQR method helps detect **unusually low or high alcohol values**.

```python
Q1 = np.percentile(alcohol, 25)
Q3 = np.percentile(alcohol, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
outliers = alcohol[
    (alcohol < lower_bound) |
    (alcohol > upper_bound)
]

print(outliers)
```
