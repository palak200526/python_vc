## IQR (Interquartile Range)

The **Interquartile Range (IQR)** measures the spread of the **middle 50% of the data**. It is also commonly used to identify **outliers**.

```python
import numpy as np
import pandas as pd
data = [10, 12, 15, 18, 20, 200, 22, 25, 28, 30, -50, 150]
```

## Calculating Q1, Q3, and IQR

This code calculates the **First Quartile (Q1)**, **Third Quartile (Q3)**, and **Interquartile Range (IQR)** of the dataset.

```python
Q1 = np.percentile(data, 25)
Q3 = np.percentile(data, 75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Detecting Outliers Using IQR

This code uses the **IQR (Interquartile Range) method** to identify outliers in the dataset.

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = [x for x in data if x < lower or x > upper]

print("Lower Bound:", lower)
print("Upper Bound:", upper)
print("Outliers:", outliers)
```

## Removing Outliers

This filters the data and removes values outside the **IQR lower and upper bounds**.

- **Original Data** → Data before removing outliers.
- **Clean Data** → Data after removing outliers.
- Values outside the bounds are removed.

```python
clean_data = [x for x in data if lower <= x <= upper]

print("Original Data:", data)
print("Clean Data:", clean_data)
```

## IQR Calculation for Salary

This code calculates **Q1, Q3, and IQR** for the salary column.

- **Q1** → 25th percentile of salaries.
- **Q3** → 75th percentile of salaries.
- **IQR** → Difference between Q3 and Q1.

### Key Idea

IQR shows the spread of the **middle 50% of salaries** and can be used to detect outliers.

```python
df = pd.DataFrame({
    "salary": [25000, 30000, 32000, 4005 ,35000, 38000,
               40000, 42000, 45000, 50000, 200000, 5000, 798600]
})

Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)

IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
```

## Detecting Salary Outliers

This code identifies **salary values that fall outside the IQR bounds**.

- **Lower Bound** → `Q1 - 1.5 × IQR`
- **Upper Bound** → `Q3 + 1.5 × IQR`
- Salaries below the lower bound or above the upper bound are treated as **outliers**.
- `outliers` contains the salary records identified as outliers.

```python
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df["salary"] < lower) |
    (df["salary"] > upper)
]

print(outliers)
```

## Detecting Outliers in Multiple Columns

This code applies the **IQR method** to every column in the DataFrame.

- Loops through each column using `df.columns`.
- Calculates **Q1, Q3, and IQR**.
- Finds the **lower and upper bounds**.
- Identifies values outside these bounds as **outliers**.
- Prints the IQR and outlier rows for each column.

### Key Idea

The same **IQR-based outlier detection** is applied automatically to both `age` and `salary`.

```python
df = pd.DataFrame({
    "age": [226, 21, 22, 40, 24, 5, 65, 37, 40, 105, 900, 32, 54],
    "salary": [25000, 30000, 32000, 4005 ,35000, 38000,
               40000, 42000, 45000, 50000, 200000, 5000, 798600]
})
for column in df.columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower) |
        (df[column] > upper)
    ]

    print(f"\nColumn: {column}")
    print("IQR:", IQR)
    print("Outliers:")
    print(outliers)
```
