## Importing Libraries

This imports the main Python libraries required for **Exploratory Data Analysis (EDA)**.

- **Pandas** → Data manipulation and analysis.
- **NumPy** → Numerical calculations and arrays.
- **Matplotlib** → Creating data visualizations.
- **Seaborn** → Creating statistical and attractive visualizations.

### Key Idea

These libraries provide the essential tools for **loading, analyzing, and visualizing data**.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Loading the Dataset

This loads the **WineQT.csv** dataset into a Pandas DataFrame.

- **`read_csv()`** → Reads the CSV file.
- **`df`** → Stores the dataset.
- **`shape`** → Returns the number of **rows and columns**.

### Key Idea

`df.shape` helps understand the **size of the dataset**.


```python
df = pd.read_csv("WineQT.csv")
df.shape
```

## Viewing Column Names

This displays the **names of all columns** present in the DataFrame.

### Key Idea

`df.columns` helps identify the **features/variables** available in the dataset.

```python
df.columns
```

## Checking Data Types

This displays the **data type of each column** in the DataFrame.

- **`int64`** → Integer values.
- **`float64`** → Decimal values.
- **`object`** → Text or categorical data.

### Key Idea

`df.dtypes` helps understand the **type of data stored in each column**.

```python
df.dtypes
```

## Filling Missing Values

This fills **missing values** in numeric columns using the **column mean**.

- **`mean()`** → Calculates the average of each numeric column.
- **`fillna()`** → Replaces missing (`NaN`) values with those averages.
- **`numeric_only=True`** → Applies the operation only to numeric columns.

### Key Idea

Missing numeric values are replaced with their **respective column mean**.

```python
df.fillna(df.mean(numeric_only=True))
```

## Alcohol Distribution

This creates a **histogram** to visualize the distribution of alcohol values.

- **20 bins** → Divides the alcohol values into 20 intervals.
- **X-axis** → Alcohol content.
- **Y-axis** → Frequency of values in each interval.
- **Title** → Describes the distribution being plotted.

### Key Idea

The histogram helps understand the **distribution and frequency of alcohol content** in the wine dataset.

```python
plt.hist(df["alcohol"], bins=20)

plt.title("Alcohol Distribution")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.show()
```

## Alcohol Boxplot

This creates a **boxplot** to visualize the distribution of alcohol values and identify potential outliers.

- **Boxplot** → Shows the median, quartiles, and spread of the data.
- **Points outside the whiskers** → Potential outliers.
- Helps understand the **range and variability** of alcohol content.

### Key Idea

A boxplot is useful for quickly detecting **outliers and the spread of numerical data**.

```python
sns.boxplot(x=df["alcohol"])

plt.title("Alcohol Boxplot")
plt.show()
```

## Alcohol vs Quality

This creates a **scatter plot** to visualize the relationship between alcohol content and wine quality.

- **X-axis** → Alcohol content.
- **Y-axis** → Wine quality.
- **Each point** → Represents one wine record.

### Key Idea

The scatter plot helps identify whether **higher alcohol content is associated with higher wine quality**.

```python
sns.scatterplot(
    data=df,
    x="alcohol",
    y="quality"
)

plt.title("Alcohol vs Quality")
plt.show()
```

## Correlation Between Alcohol and Quality

This calculates the **correlation** between alcohol content and wine quality.

- **`corr()`** → Measures the relationship between two numerical variables.
- **Positive value** → Both tend to increase together.
- **Negative value** → One tends to increase while the other decreases.
- **Near 0** → Weak or no linear relationship.

### Key Idea

It helps determine the **strength and direction of the linear relationship** between alcohol and quality.

```python
df["alcohol"].corr(df["quality"])
```

## Correlation Heatmap

This creates a **correlation heatmap** for all numerical columns in the dataset.

- **Correlation values** → Show the strength and direction of relationships.
- **`annot=True`** → Displays the correlation values inside the cells.
- **`coolwarm`** → Uses different colors to represent positive and negative correlations.
- **`numeric_only=True`** → Includes only numerical columns.

### Key Idea

The heatmap helps quickly identify **strong, weak, positive, and negative relationships** between variables.

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Pair Plot

This creates a **pair plot** to visualize relationships between selected numerical variables.

- Shows **scatter plots** between pairs of variables.
- Shows the **distribution** of each variable.
- Helps identify **patterns, trends, and correlations**.
- Includes `alcohol`, `pH`, `sulphates`, `volatile acidity`, and `quality`.

### Key Idea

A pair plot provides a **quick overview of relationships between multiple variables**.

```python
sns.pairplot(
    df[[
        "alcohol",
        "pH",
        "sulphates",
        "volatile acidity",
        "quality"
    ]]
)

plt.show()
```
