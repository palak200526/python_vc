## Normal Distribution Properties
This imports the libraries such as numpy, matplotlib.pyplot, seaborn

````python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
````

**This code generates 1,000 random values from a normal (Gaussian) distribution with:**

- **Mean (`loc`)** = 50
- **Standard deviation (`scale`)** = 10
- **Number of values (`size`)** = 1000

````python
data = np.random.normal(loc=50, scale=10, size=1000)
````

**This code creates a histogram to visualize the distribution of the generated data.**

- `plt.hist(data, bins=30)` → Creates a histogram using **30 bins**.
- `plt.title()` → Sets the title to **Normal Distribution**.
- `plt.xlabel()` → Labels the x-axis as **Value**.
- `plt.ylabel()` → Labels the y-axis as **Frequency**.
- `plt.show()` → Displays the histogram.

````python
plt.hist(data, bins=30)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()
````

**This code creates a histogram along with a KDE (Kernel Density Estimate) curve to visualize the shape and density of the normal distribution.**

- `sns.histplot(data, kde=True)` → Creates a histogram and adds a **KDE curve**.
- `plt.title()` → Sets the title to **Normal Distribution with KDE**.
- `plt.xlabel()` → Labels the x-axis as **Value**.
- `plt.ylabel()` → Labels the y-axis as **Density**.
- `plt.show()` → Displays the plot.

````python
sns.histplot(data, kde=True)

plt.title("Normal Distribution with KDE")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
````

## Mean and Standard Deviation

The following code calculates the **mean** and **standard deviation** of the dataset.

- **Mean** → Represents the average value of the data.
- **Standard Deviation** → Measures how much the data values are spread out from the mean.

```python
mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std)

````python
mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std)
````

## 68-95-99.7 Rule (Empirical Rule)

The **68-95-99.7 rule**, also called the **Empirical Rule**, describes how data is distributed around the mean in a **normal distribution**.

- **Within 1 Standard Deviation (SD)** → approximately **68%** of the data
- **Within 2 Standard Deviations (SD)** → approximately **95%** of the data
- **Within 3 Standard Deviations (SD)** → approximately **99.7%** of the data

### Code Explanation

```python
within_1 = np.sum((data >= mean - std) & (data <= mean + std))
within_2 = np.sum((data >= mean - 2*std) & (data <= mean + 2*std))
within_3 = np.sum((data >= mean - 3*std) & (data <= mean + 3*std))

````python
# 68-95-99.7 rule (Empirical rule)
within_1 = np.sum((data >= mean - std) & (data <= mean + std))
within_2 = np.sum((data >= mean - 2*std) & (data <= mean + 2*std))
within_3 = np.sum((data >= mean - 3*std) & (data <= mean + 3*std))

print("Within 1 SD:", within_1 / len(data) * 100, "%")
print("Within 2 SD:", within_2 / len(data) * 100, "%")
print("Within 3 SD:", within_3 / len(data) * 100, "%")
````

## Comparing Different Standard Deviations

This code demonstrates how changing the **standard deviation (SD)** affects the shape and spread of a normal distribution.

```python
data1 = np.random.normal(50, 5, 1000)
data2 = np.random.normal(50, 10, 1000)
data3 = np.random.normal(50, 20, 1000)

sns.kdeplot(data1, label="SD = 5")
sns.kdeplot(data2, label="SD = 10")
sns.kdeplot(data3, label="SD = 20")

plt.title("Effect of Standard Deviation")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()

````python
# comparing different Standard Deviation
data1 = np.random.normal(50, 5, 1000)
data2 = np.random.normal(50, 10, 1000)
data3 = np.random.normal(50, 20, 1000)

sns.kdeplot(data1, label="SD = 5")
sns.kdeplot(data2, label="SD = 10")
sns.kdeplot(data3, label="SD = 20")

plt.title("Effect of Standard Deviation")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.show()
````
