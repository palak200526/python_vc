## Creating a Contour Plot

This creates a contour plot for the function **Z = X² + Y²**.

- **`linspace`** → Creates 100 values between -5 and 5 for X and Y.
- **`meshgrid`** → Creates a 2D grid of X and Y values.
- **Z** → Calculates the value of the function at each grid point.
- **`contour`** → Draws contour lines representing different Z values.
- **Labels and title** → Add names to the axes and plot.

### Key Idea

The contour lines show points where **Z has the same value**.

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)

X, Y = np.meshgrid(x, y)

Z = X**2 + Y**2

plt.contour(X, Y, Z)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Contour Plot")

plt.show()
```

## Filled Contour Plot

This creates a **filled contour plot**, where different colors represent different **Z values**.

- **`contourf`** → Fills the contour regions with colors.
- **Colorbar** → Shows the relationship between colors and Z values.
- **X-axis** → X variable.
- **Y-axis** → Y variable.
- **Colors** → Represent different Z values.

### Key Idea

`contour` shows **lines**, while `contourf` shows **filled colored regions**.

```python
plt.contourf(X, Y, Z)

plt.colorbar()

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Filled Contour Plot")

plt.show()
```

## Contour Plot with Labels

This creates a **contour plot with 10 contour levels** and labels each line with its Z value.

- **`levels=10`** → Creates 10 contour levels.
- **`clabel()`** → Displays the Z values directly on the contour lines.
- **X-axis** → X variable.
- **Y-axis** → Y variable.

### Key Idea

The labels make it easier to understand the **Z value represented by each contour line**.

```python
contours = plt.contour(X, Y, Z, levels=10)

plt.clabel(contours)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Contour Plot with Labels")

plt.show()
```
