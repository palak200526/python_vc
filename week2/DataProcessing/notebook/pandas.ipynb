## Creating a Pandas Series

This creates a **Pandas Series** containing the values `10, 20, 30, 40, 50`.

- **Series** → A one-dimensional labeled data structure.
- Pandas automatically assigns **index numbers** starting from `0`.
- The values are stored along with their corresponding indexes.

### Key Idea

A **Series** is similar to a single column in a table.

```python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])
print(data)
```

## Creating a Pandas DataFrame

This creates a **Pandas DataFrame** using a dictionary containing student information.

- **DataFrame** → A two-dimensional table with rows and columns.
- **Columns** → `Name`, `Age`, and `Marks`.
- Each dictionary key becomes a **column name**.
- Each list contains the values for that column.
- Pandas automatically creates **row indexes** starting from `0`.

### Key Idea

A **DataFrame** is similar to a table in Excel or a database.

```python
import pandas as pd

data = {
    "Name": ["Aman", "Riya", "Rahul"],
    "Age": [20, 21, 22],
    "Marks": [85, 90, 78]
}

df = pd.DataFrame(data)

print(df)
```

## Grouping and Calculating Mean

This calculates the **average fixed acidity** for each **quality level**.

- **`groupby("quality")`** → Groups the data by quality.
- **`["fixed acidity"]`** → Selects the fixed acidity column.
- **`mean()`** → Calculates the average for each group.

### Key Idea

It shows the **average fixed acidity for each wine quality level**.

```python
df.groupby("quality")["fixed acidity"].mean()
```

## Grouping and Aggregating Data

This calculates multiple statistics for **alcohol** for each **quality level**.

- **`groupby("quality")`** → Groups the data by quality.
- **`mean`** → Average alcohol value.
- **`min`** → Lowest alcohol value.
- **`max`** → Highest alcohol value.

### Key Idea

It gives a **summary of alcohol content for each wine quality level**.

```python
df.groupby("quality")["alcohol"].agg(["mean", "min", "max"])
```
