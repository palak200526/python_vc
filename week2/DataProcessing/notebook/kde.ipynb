## Basic KDE Plot with Seaborn

This example demonstrates how to create a simple Kernel Density Estimation (KDE) plot using **Seaborn**:

- **Data**: A small list of numerical values.
- **KDE**: Seaborn’s `kdeplot` function estimates the probability density function of the data.
- **Visualization**: The plot shows a smooth curve representing the distribution of values.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

data = [10, 12, 15, 15, 18, 20, 21, 22, 25, 25, 28, 30]

sns.kdeplot(data=data)

plt.title("Basic KDE Plot")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

## Histogram with KDE

This plot combines a histogram with a Kernel Density Estimation (KDE) curve:
- The **histogram** shows the frequency of data values in bins.
- The **KDE curve** overlays a smooth estimate of the distribution.
- Together, they provide both discrete counts and a continuous density view of the dataset.

```python
sns.histplot(data, kde=True)

plt.title("Histogram with KDE")
plt.show()
```

## Distribution of Marks

This plot uses Kernel Density Estimation (KDE) to visualize the distribution of student marks:
- The **smooth curve** shows how marks are spread across the dataset.
- Peaks in the curve indicate ranges where marks are more concentrated.
- KDE provides a continuous view of the distribution, making it easier to spot trends compared to raw data points.


```python
df = pd.DataFrame({
    "marks": [45, 50, 52, 55, 58, 60, 62, 65, 68, 70,
              72, 75, 78, 80, 82, 85, 88, 90]
})

sns.kdeplot(data=df, x="marks")

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Density")
plt.show()
```

##  Sepal Length Distribution by Species

This KDE plot shows how **sepal length** varies across different iris species:
- Each species is represented with a separate density curve using the `hue` parameter.
- The curves highlight differences in typical sepal lengths between species.
- Overlapping regions indicate similarities, while distinct peaks reveal unique traits.


```python
df = sns.load_dataset("iris")

sns.kdeplot(data=df, x="sepal_length", hue="species")

plt.title("Sepal Length Distribution by Species")
plt.show()
```

## KDE with Different Bandwidths

This set of plots shows how the **bandwidth parameter** affects the smoothness of the KDE curve:

- **Bandwidth = 0.3** → Very detailed curve, sensitive to individual data points.  
  Captures fine variations but may appear jagged.

- **Bandwidth = 1** → Balanced smoothness.  
  Provides a good compromise between detail and readability.

- **Bandwidth = 2** → Very smooth curve.  
  General trends are clear, but smaller peaks and variations are flattened.

Bandwidth tuning is essential: too small can overfit noise, too large can oversmooth and hide important features.


```python
sns.kdeplot(data=data, bw_adjust=0.3)

plt.title("KDE with 0.3 Bandwidth")
plt.show()
```

```python
sns.kdeplot(data=data, bw_adjust=1)

plt.title("KDE with 1 Bandwidth")
plt.show()
```

```python
sns.kdeplot(data=data, bw_adjust=2)

plt.title("KDE with 2 Bandwidth")
plt.show()
```

## Distribution of Total Bill

This KDE plot shows the distribution of restaurant bills from the **Tips dataset**:
- The smooth curve represents how total bill amounts are spread across customers.

## KDE with Mean and Median

This plot shows the distribution of restaurant bills using Kernel Density Estimation (KDE), with two reference lines:
- **Mean (average)**: Indicates the central tendency of all bill amounts.
- **Median (middle value)**: Shows the point where half the bills are below and half are above.

Comparing the mean and median helps identify skewness:
- If the mean is greater than the median → distribution is right-skewed (long tail on higher bills).
- If the mean is less than the median → distribution is left-skewed (long tail on lower bills).

```python
mean = df["total_bill"].mean()
median = df["total_bill"].median()

sns.kdeplot(data=df, x="total_bill")

plt.axvline(mean, linestyle="--", label=f"Mean = {mean:.2f}")
plt.axvline(median, linestyle="--", label=f"Median = {median:.2f}")

plt.legend()
plt.title("KDE with Mean and Median")
plt.show()
```
